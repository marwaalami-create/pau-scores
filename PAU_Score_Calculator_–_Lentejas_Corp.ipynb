```python

# PAU SCORE CALCULATOR
# Lentejas Corp.


# 1. MARWA - DATA INPUT

def input_student_data() -> tuple:
    """
    Ask the student for their grades.

    Returns:
        tuple: Student data and subject weights.
    """

    marks_PAU = {}

    general_subjects = [
        "Mathematics II",
        "Spanish and Literature II",
        "Spain History",
        "English II"
    ]

    optional_subjects = [
        "Physics",
        "Technology II",
        "Chemistry",
        "Technical drawing II"
    ]

    ponderaciones = {
        "Mathematics II": 0.2,
        "Physics": 0.2,
        "Technology II": 0.2,
        "Chemistry": 0.1,
        "Technical drawing II": 0.1
    }

    print("PAU SCORE CALCULATOR")

    student_type = input(
        "Bachillerato (B) or Superior Cycle (C): "
    ).upper()

    while student_type != "B" and student_type != "C":
        student_type = input("Enter B or C: ").upper()

    if student_type == "B":

        first = float(input("First year average: "))
        second = float(input("Second year average: "))

        for subject in general_subjects:
            grade = float(input(subject + ": "))
            marks_PAU[subject] = grade

        academic_record = (
            first,
            second,
            general_subjects
        )

    else:
        average = float(input("Cycle average: "))
        academic_record = (average,)

    for subject in optional_subjects:

        grade = input(subject + " (Enter to skip): ")

        if grade != "":
            marks_PAU[subject] = float(grade)

    student_data = (marks_PAU, academic_record)

    return student_data, ponderaciones


# 2. CLAUDIA - ADMISSION SCORE

def compute_admission_score(scores: tuple[dict, tuple], ponderaciones: dict) -> float:
    """
    Calculate the final PAU university admission score (out of 14).

    Combines the access score and the option score.
    Returns 0.0 if the student does not pass.
    """

    marks_pau_exams, academic_record = scores

    # Obtain the access score
    access, passed = access_score(scores)

    # Check the minimum requirements
    if not passed or access < 5.0:
        return 0.0

    # Obtain the option score
    option_score = compute_option_score(marks_pau_exams, ponderaciones)

    # Add both scores and round the result
    admission_score = round(access + option_score, 3)

    return admission_score


# 3. BESAY - ACCESS SCORE

def access_score(student_data: tuple[dict, tuple]) -> tuple[float, bool]:
    """
    Calculate the access score of the student.

    Returns:
        tuple: Access score and pass status.
    """

    p = True
    mark = float(0)

    academic_record = student_data[1]
    marks_PAU = student_data[0]

    # Student from Superior Cycle
    if len(academic_record) == 1:

        mark = float(academic_record[0])

        result = (mark, p)

        return result

    # Student from Bachillerato
    else:

        bachelor_average = (academic_record[0] + academic_record[1]) / 2

        # Calculate general phase using Yassin's function
        general_phase_score = compute_general_phase(marks_PAU, academic_record[2])

        if general_phase_score >= 4:
            p = True
        else:
            p = False

        final_mark = (0.6 * bachelor_average + 0.4 * general_phase_score)

        result = (final_mark, p)

        return result


# 4. YASSIN - GENERAL PHASE

def compute_general_phase(exam_scores: dict, subjects_general_phase: list) -> float:
    """
    Calculate the average grade of the general PAU phase.

    Args:
        exam_scores (dict): Dictionary containing the exam grades.
        subjects_general_phase (list): List of general phase subjects.

    Returns:
        float: The average grade of the general PAU phase.
    """

    total = 0

    for subject in subjects_general_phase:
        total += exam_scores[subject]

    return total / len(subjects_general_phase)


# 5. LAURA - OPTION SCORE

def compute_option_score(exam_scores: dict, weights: dict) -> float:
    """
    Returns the highest possible score (out of 4)
    for the option phase of the PAU exam.

    Args:
        exam_scores: Dictionary with the exam grades.
        weights: Dictionary with subject weights.

    Returns:
        float: Sum of the two best weighted scores.
    """

    if len(exam_scores) == 0:
        return 0.0

    # Filter invalid subjects
    filtered = filter_option_scores(exam_scores, weights)

    weighted_scores = []

    for subject in filtered:
        weighted_scores.append(float(filtered[subject] * weights[subject]))

    # Return the best two weighted scores
    return (
        sum(sorted(weighted_scores, reverse=True)[:2])
        if len(weighted_scores) > 2
        else sum(weighted_scores)
    )


# 6. GABRIELA - FILTER OPTION SCORES

def filter_option_scores(marks_PAU_exams: dict, ponderable: dict) -> dict:
    """
    Filter valid optional subjects.

    Args:
        marks_PAU_exams: Dictionary with exam grades.
        ponderable: Dictionary with subject weights.

    Returns:
        dict: Subjects with valid grades and weights.
    """

    filtered_scores = {}

    for subject, score in marks_PAU_exams.items():

        if subject in ponderable:

            if score >= 5.0 and ponderable[subject] >= 0.1:

                filtered_scores[subject] = score

    return filtered_scores


# MAIN PROGRAM

student_data, ponderable = input_student_data()

final_score = compute_admission_score(
    student_data,
    ponderable
)

print("------------------------------")
print("FINAL PAU SCORE:", final_score)
print("------------------------------")
```
